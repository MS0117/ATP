```python
from vllm import LLM, SamplingParams

# 모델 이름 설정
model_name = "deepseek-ai/DeepSeek-Prover-V1.5-Base"

# vLLM을 사용하여 모델 로드
llm = LLM(model_name, dtype="float16",max_model_len=512)  # float16 사용 (메모리 절약)

# Sampling 설정
sampling_params = SamplingParams(
    temperature=0.7,  # 다양성 조절 (낮으면 보수적, 높으면 창의적)
    top_p=0.9,  # nucleus sampling
    max_tokens=256,  # 최대 생성 길이
)

# 프롬프트 입력
prompt = "Theorem: The sum of two even numbers is always even.\nProof:"

# 모델에 입력 후 생성
outputs = llm.generate(prompt, sampling_params)

# 결과 출력
for output in outputs:
    print("Generated Text:\n", output.outputs[0].text)
print("prompt",prompt)
print("outputs",output.outputs[0].text)
```

```python
full_text=prompt+completions_text[0]
print(full_text)
```

```python
full_encoded = tokenizer(
            full_text,
            return_offsets_mapping=True,
            add_special_tokens=False
        )
```

```python
input_ids = full_encoded["input_ids"]
offsets = full_encoded["offset_mapping"]  # list of (start, end)
print(input_ids)
```

```python
token_scores = []
token_texts = []
prompt_len = len(prompt)
token_id=[]
for tid, (start, end) in zip(input_ids, offsets):           #character-token alignment
    # If you only want tokens fully in the completion portion,
    # check if start >= prompt_len
    # (If you want partial coverage for tokens that straddle the boundary,
    # you'd have to do a partial average or something more advanced.)
    if start >= prompt_len:

        # The raw text for this token
        token_str = full_text[start:end]
        token_scores.append(1)
        token_texts.append(token_str)
        token_id.append(tid)
```
