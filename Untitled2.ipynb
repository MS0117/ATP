```python
import torch
from transformers import AutoTokenizer

model_name = "deepseek-ai/DeepSeek-Prover-V1.5-Base"
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)

# Check BOS and EOS token info
bos_token = tokenizer.bos_token  # BOS token string
bos_token_id = tokenizer.bos_token_id  # BOS token ID
eos_token = tokenizer.eos_token  # EOS token string
eos_token_id = tokenizer.eos_token_id  # EOS token ID

print(f"BOS Token: {bos_token}, BOS Token ID: {bos_token_id}")
print(f"EOS Token: {eos_token}, EOS Token ID: {eos_token_id}")

# Sample text
text = "Hello, how are you?"

# Encode with and without special tokens
encoded_with_special = tokenizer.encode(text, add_special_tokens=True)
encoded_without_special = tokenizer.encode(text, add_special_tokens=False)

print("Encoded with special tokens:", encoded_with_special)
print("Encoded without special tokens:", encoded_without_special)

# Decode to check if EOS appears in text form
decoded_with_special = tokenizer.decode(encoded_with_special)
decoded_without_special = tokenizer.decode(encoded_without_special)

print("Decoded with special tokens:", decoded_with_special)
print("Decoded without special tokens:", decoded_without_special)

# Check if BOS and EOS token IDs are present in the encoded sequence
contains_bos = bos_token_id in encoded_with_special if bos_token_id is not None else False
contains_eos = eos_token_id in encoded_with_special if eos_token_id is not None else False

print(f"Does the encoded sequence include BOS? {contains_bos}")
print(f"Does the encoded sequence include EOS? {contains_eos}")
```
